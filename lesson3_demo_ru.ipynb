## 1. Кортежи (tuple)

Упорядоченная **неизменяемая** коллекция. Идеальна для фиксированных наборов: координаты, RGB, дата.

```python
capitals = ("Москва", "Ташкент", "Бишкек")
point = (10, 20)
single = (42,)       # запятая обязательна! иначе это просто int в скобках

print(type(single))  # <class 'tuple'>
print(type((42)))    # <class 'int'>

# распаковка
x, y = point
print(x, y)
```

```python
# попытка изменить — ошибка
try:
    capitals[0] = "X"
except TypeError as e:
    print("TypeError:", e)
```

## 2. Множества (set)

Неупорядоченная коллекция **уникальных** элементов. Главные суперспособности: дедупликация и операции теории множеств.

```python
tags = {"python", "sql", "python", "git"}
print(tags)            # дубли исчезли

# проверка принадлежности — O(1), очень быстро
print("python" in tags)

# дедупликация списка
items = [1, 2, 2, 3, 4, 3, 5, 1]
print(list(set(items)))  # порядок может потеряться
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("объединение:", a | b)
print("пересечение:", a & b)
print("разность:",    a - b)
print("симметр.:",    a ^ b)
```

## 3. Словари (dict)

Пары **ключ → значение**. Чтение по ключу — O(1).

```python
user = {"name": "Алиса", "age": 30}

# чтение
print(user["name"])

# [k] на отсутствующем ключе — KeyError
try:
    print(user["city"])
except KeyError as e:
    print("KeyError:", e)

# безопасно через .get()
print(user.get("city"))        # None
print(user.get("city", "NA"))  # NA
```

```python
# модификация
user["city"] = "Бухара"   # добавить
user["age"] = 31           # обновить
del user["age"]            # удалить
print(user)

# три способа перебора
scores = {"Аня": 85, "Боря": 72, "Вика": 91}

for k in scores:                # только ключи
    print(k, end=" ")
print()
for v in scores.values():       # только значения
    print(v, end=" ")
print()
for k, v in scores.items():     # пары
    print(f"{k}={v}", end=" ")
```

## 4. def · return

Функция — кухонный комбайн: ингредиенты входят, результат выходит через `return`.

```python
def greet(name):
    """Приветствует пользователя."""
    return f"Привет, {name}!"

print(greet("Алиса"))

# несколько значений → кортеж + распаковка
def divmod_safe(a, b):
    if b == 0:
        return None
    return a // b, a % b

q, r = divmod_safe(17, 5)
print(q, r)
```

## 5. Параметры по умолчанию

Если аргумент не передали — берётся default.

```python
def greet(name, greeting="Привет"):
    return f"{greeting}, {name}!"

print(greet("Аня"))                       # default "Привет"
print(greet("Боря", "Салом"))             # переопределили
print(greet(name="Вика", greeting="Hi"))  # именованные
```

### 🔥 Ловушка: Mutable Default Arguments

Default вычисляется **один раз** — при объявлении функции. Если default — список или словарь, все вызовы делят один объект.

```python
# ❌ опасный код
def add_to_cart(item, cart=[]):
    cart.append(item)
    return cart

print(add_to_cart("яблоко"))   # ['яблоко']
print(add_to_cart("груша"))    # ['яблоко', 'груша']  ← внезапно!
print(add_to_cart("банан"))    # ['яблоко', 'груша', 'банан']

# смотрим на default — это тот же объект
print("id(default):", id(add_to_cart.__defaults__[0]))
```

```python
# ✓ правильно
def add_to_cart(item, cart=None):
    if cart is None:
        cart = []   # новый список на каждый вызов
    cart.append(item)
    return cart

print(add_to_cart("яблоко"))
print(add_to_cart("груша"))
print(add_to_cart("банан"))
```

## 6. *args и **kwargs

* `*args` — позиционные «лишние» аргументы → tuple
* `**kwargs` — именованные → dict

```python
def total(*args):
    print("args =", args, type(args).__name__)
    return sum(args)

print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())  # 0 — sum пустого

def profile(**kwargs):
    print("kwargs =", kwargs, type(kwargs).__name__)
    for k, v in kwargs.items():
        print(f"  {k}: {v}")

profile(name="Аня", age=30, city="Ташкент")
```

### 🎯 Квиз: универсальная сигнатура

Порядок строгий: `обычные → defaults → *args → keyword-only → **kwargs`.

**Перед запуском — попросите студентов угадать, куда попадёт каждое значение.**

```python
def f(a, b=10, *args, status="active", **kwargs):
    print("a       =", a)
    print("b       =", b)
    print("args    =", args)
    print("status  =", status)
    print("kwargs  =", kwargs)

f(1, 2, 3, 4, status="pending", user="John")

# Ожидаемо:
#   a = 1
#   b = 2     (перезаписал default 10)
#   args = (3, 4)
#   status = "pending"  (keyword-only — нельзя позиционно после *args)
#   kwargs = {"user": "John"}
```

### Распаковка при вызове

Используем `*` и `**` чтобы **разложить** list/dict в аргументы.

```python
nums = [1, 2, 3, 4, 5]
print(total(*nums))   # эквивалентно total(1, 2, 3, 4, 5)

info = {"name": "Вика", "age": 28}
profile(**info)       # эквивалентно profile(name="Вика", age=28)
```

## 7. Область видимости (scope) + global / nonlocal

Правило **LEGB**: Local → Enclosing → Global → Built-in.

```python
x = 100  # global

def demo():
    y = 5            # local
    print("вижу x =", x, "и y =", y)

demo()
try:
    print(y)   # y живёт только в demo()
except NameError as e:
    print("NameError:", e)
```

### UnboundLocalError — самая частая ошибка scope

Как только Python видит **присваивание** имени в функции — он считает имя локальным.

```python
counter = 0

def inc_broken():
    counter += 1   # Python считает counter локальной
    return counter

try:
    inc_broken()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)
```

```python
# ✓ фикс через global
counter = 0

def inc():
    global counter
    counter += 1
    return counter

inc(); inc(); inc()
print("counter =", counter)
```

```python
# nonlocal — для вложенных функций (замыкания)
def make_counter():
    n = 0
    def inc():
        nonlocal n
        n += 1
        return n
    return inc

c = make_counter()
print(c(), c(), c())   # 1 2 3
```

## 8. lambda — короткая анонимная функция

```python
double = lambda x: x * 2
print(double(5))

people = [("Аня", 30), ("Боря", 25), ("Вика", 28), ("Глеб", 22), ("Дима", 35)]

# сортировка по возрасту
print(sorted(people, key=lambda p: p[1]))

# сортировка по имени
print(sorted(people, key=lambda p: p[0]))

# фильтр взрослых
print(list(filter(lambda p: p[1] >= 28, people)))

# только имена
print(list(map(lambda p: p[0], people)))
```

### ⚠ Когда lambda — это плохо

Если в лямбде появляются условия / разветвления — переписывайте в обычный `def`. **Код пишется для людей.**

```python
# ❌ ужасно — три вложенных тернара в одной строке
group = lambda x: "Взрослый" if x >= 18 else ("Подросток" if x >= 13 else "Ребёнок")

print(group(25), group(15), group(8))


# ✓ читаемо
def age_group(age):
    if age >= 18: return "Взрослый"
    if age >= 13: return "Подросток"
    return "Ребёнок"

print(age_group(25), age_group(15), age_group(8))
```

### 🔥 Ловушка: Late Binding в lambda

Лямбда захватывает **имя**, а не значение в момент создания.

```python
# ❌ ожидаем 0, 1, 2 — получим 2, 2, 2
funcs = [lambda: i for i in range(3)]
for f in funcs:
    print(f())
```

```python
# ✓ фикс — «защёлкнуть» текущее i через default-параметр
funcs = [lambda i=i: i for i in range(3)]
for f in funcs:
    print(f())
```

## 9. Финал — собираем всё вместе

Топ-N студентов: dict + lambda + sorted + распаковка.

```python
def top_n(scores, n=3, *, ascending=False):
    pairs = sorted(scores.items(), key=lambda p: p[1], reverse=not ascending)
    return pairs[:n]

scores = {"Аня": 85, "Боря": 72, "Вика": 91, "Глеб": 68, "Дима": 88}

print("Топ-3 лучших:")
for name, score in top_n(scores, 3):
    print(f"  {name}: {score}")

print("\nТоп-2 худших:")
for name, score in top_n(scores, 2, ascending=True):
    print(f"  {name}: {score}")
```
